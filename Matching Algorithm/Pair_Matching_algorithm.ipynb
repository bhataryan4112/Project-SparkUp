```python
import pandas as pd
import numpy as np
import os
from datetime import datetime
import pytz  # Import pytz for timezone handling

# Replace 'your_file.csv' with the actual name of your CSV file
file_name = '/content/ProjectSparkup - Sheet1.csv'
print(file_name)
try:
    df = pd.read_csv(file_name)
except FileNotFoundError:
    print(f"Error: File '{file_name}' not found.")
    exit()
except pd.errors.ParserError:
    print(f"Error: Could not parse the CSV file '{file_name}'. Check its format.")
    exit()
except Exception as e:
    print(f"An unexpected error occurred: {e}")
    exit()

columns_to_drop = [
    'Name', 'Mobile No.', 'Age', 'Year', 'Campus', 'Gender', 'Question1', 'Question2',
    'Question3', 'Question4', 'Question5', 'Question6', 'Question7', 'Question8',
    'Question9', 'Question10', 'Question11', 'Question12', 'Question13', 'Question14',
    'Question15', 'Question16', 'Question17', 'Question18', 'Question19', 'Question20',
    'Fun Fact', 'timestamp', 'timestamp.1'
]

# Drop the specified columns
df = df.drop(columns=columns_to_drop, errors='ignore')

# Create an empty list to store all matches
all_matches = []

def createPairs(email):
    # Get the Big 5 characteristics for the given email
    current_person = df[df[email_column] == email]
    if current_person.empty:
        print(f"Email '{email}' not found in the DataFrame.")
        return []

    current_characteristics = current_person[
        ['Openness', 'Conscientiousness', 'Extraversion', 'Agreeableness', 'Neuroticism']
    ].values[0]

    # Calculate distances to all other people, considering complementarity
    distances = []
    for index, row in df.iterrows():
        if row[email_column] != email:  # Exclude the current person
            other_characteristics = row[
                ['Openness', 'Conscientiousness', 'Extraversion', 'Agreeableness', 'Neuroticism']
            ].values

            # Calculate similarity distance (Euclidean distance)
            similarity_distance = np.linalg.norm(current_characteristics - other_characteristics)

            # Calculate complementarity distance (absolute difference on specific traits)
            complementarity_distance = np.sum(np.abs(current_characteristics - other_characteristics))

            # Combine similarity and complementarity (adjust weights as needed)
            combined_distance = 0.7 * similarity_distance + 0.3 * complementarity_distance

            distances.append((row[email_column], combined_distance))

    # Sort by combined distance and get the top 5 closest emails
    distances.sort(key=lambda item: item[1])  # Sort by combined_distance
    top_5_emails = [email for email, _ in distances[:5]]  # Get top 5 emails based on combined_distance

    # Store the top 5 matches in a dictionary
    matches_dict = {
        'Email': email,
        'Match 1': top_5_emails[0] if len(top_5_emails) > 0 else None,
        'Match 2': top_5_emails[1] if len(top_5_emails) > 1 else None,
        'Match 3': top_5_emails[2] if len(top_5_emails) > 2 else None,
        'Match 4': top_5_emails[3] if len(top_5_emails) > 3 else None,
        'Match 5': top_5_emails[4] if len(top_5_emails) > 4 else None
    }

    # Append the dictionary to the all_matches list
    all_matches.append(matches_dict)

    # Print the top 5 matches and their distances (optional)
    print(f"Top 5 matches for {email}:")
    for email_match, distance in distances[:5]:
        print(f" - {email_match}: {distance}")

    return top_5_emails

# Iterate through the emails in the specified column
email_column = "Email Id"  # Replace 'Name' with your actual column name containing emails

if email_column in df.columns:
    for email in df[email_column]:
        if pd.notna(email):  # Check if the email is not NaN
            createPairs(email)

    # Create a DataFrame from all_matches
    all_matches_df = pd.DataFrame(all_matches)

    # Get current date and time for filename with timezone
    current_datetime = datetime.now(pytz.timezone('Asia/Kolkata')).strftime("%Y-%m-%d_%H-%M-%S")  # Replace 'YOUR_TIMEZONE'
    filename = f"{current_datetime}-matches.csv"

    # Write the DataFrame to the CSV file
    all_matches_df.to_csv(filename, index=False)

else:
    print(f"Error: Column '{email_column}' not found in the DataFrame.")
```
